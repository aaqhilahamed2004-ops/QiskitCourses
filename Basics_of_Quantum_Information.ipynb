```python

from qiskit.quantum_info import Operator

Y = Operator([[0, -1.0j], [1.0j, 0]])
X = Operator([[0, 1], [1, 0]])
H = Operator([[1/np.sqrt(2), 1/np.sqrt(2)], [1/sqrt(2), -1/sqrt(2)]])
S = Operator([[1, 0], [0, 1.0j]])
T = Operator([[1, 0], [0, (1+1.0j)/sqrt(2)]])
```

## Ordering of Qubits in Qiskit

Qiskit normally follows the little-endian order.

`|most-significant-qubit⟩......|least-significant-qubit⟩`

### Statevector function

`Statevector([0, 1, 2, 3, 4, 5, 6, 7])`

It assigns values in this indexing order.

So, for a `phi = Statevector([1, 1, 0, 0]/sqrt(2))`
the Statevector would be:

$$phi = \frac{1}{\sqrt2}|00⟩ + \frac{1}{\sqrt2} |10⟩$$

###What does `.measure()` do when given multiple indices?

`.meausre()` normally gives two outputs.
  1. `outcome` - the measured value (0 or 1 for single qubits).
  2.`state` - the state after measurement.

But when given arguments - precise arguments as to which index/indices to measure, the order of the `outcome` is as follow:

for measure([2,0])

it measures the 2nd indexed qubit first. shows the result down.
then, measures the 0th indexed qubit and writes the value down, pushing the 2nd qubit value to the right.
